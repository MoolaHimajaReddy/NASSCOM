```python
# Core imports for the whole lab
import numpy as np
import sympy as sp

x, y = sp.symbols('x y')      # symbolic variables we'll reuse
sp.init_printing()            # pretty-print symbolic math
np.random.seed(42)
print('Setup complete. SymPy', sp.__version__, '| NumPy', np.__version__)
```

```python
# -----------------------------------------------------------
# 🔹 1A. NUMERICAL DERIVATIVE (finite difference)
# -----------------------------------------------------------

# The derivative is the slope: how much f changes for a tiny step h
def f(x):
    return x**2

h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h
print('Numerical f\'(3):', round(slope_at_3, 4), ' (exact = 6)')
```

```python
# -----------------------------------------------------------
# 🔹 1B. SYMBOLIC DERIVATIVE (SymPy)
# -----------------------------------------------------------

expr = x**2
deriv = sp.diff(expr, x)          # differentiate w.r.t. x
print('d/dx (x**2) =', deriv)     # 2*x

# Evaluate the symbolic derivative at x = 3
print('Symbolic f\'(3) =', deriv.subs(x, 3))
```

```python
import sympy as sp


# Function definition
def g(x):
    return x**3 + 2 * x


# 1. Numerical derivative at x = 2 (using central difference: (g(x+h) - g(x-h)) / (2*h))
x_val = 2.0
h = 1e-6
num_diff = (g(x_val + h) - g(x_val - h)) / (2 * h)

print(f"1. Numerical derivative at x = 2: {num_diff:.6f}")
print("-" * 45)

# 2. Symbolic derivative of x**3 + 2*x
x_sym = sp.Symbol("x")
g_sym = x_sym**3 + 2 * x_sym
sym_diff_expr = sp.diff(g_sym, x_sym)

print("2. Symbolic derivative expression:")
print("g'(x) =", sym_diff_expr)
print("-" * 45)

# 3. Evaluate the symbolic derivative at x = 2 and compare
sym_diff_val = float(sym_diff_expr.subs(x_sym, x_val))

print("3. Comparison:")
print(f"Symbolic derivative at x = 2:  {sym_diff_val}")
print(f"Numerical derivative at x = 2: {num_diff:.6f}")
print("Do they match?:", abs(sym_diff_val - num_diff) < 1e-5)
```

```python
# -----------------------------------------------------------
# 🔹 2A. PARTIAL DERIVATIVES
# -----------------------------------------------------------

f2 = x**2 + 3*x*y + y**2

# A partial derivative differentiates ONE variable, holding others fixed
print('df/dx =', sp.diff(f2, x))     # 2*x + 3*y
print('df/dy =', sp.diff(f2, y))     # 3*x + 2*y
```

```python
# -----------------------------------------------------------
# 🔹 2B. THE GRADIENT (vector of partials)
# -----------------------------------------------------------

# The gradient stacks every partial derivative into one vector
grad = [sp.diff(f2, v) for v in (x, y)]
print('grad f =', grad)

# Evaluate the gradient at the point (x=1, y=2)
grad_at = [g.subs({x: 1, y: 2}) for g in grad]
print('grad f at (1, 2) =', grad_at)   # points in the steepest-ascent direction
```

```python
import sympy as sp

# Define symbolic variables
x, y = sp.symbols("x y")

# Function definition
h2 = x**2 * y + sp.sin(y)

# 1. Partial derivatives dh/dx and dh/dy
dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)

print("1. Partial Derivatives:")
print("dh/dx =", dh_dx)
print("dh/dy =", dh_dy)
print("-" * 40)

# 2. Assemble the gradient list
grad_h2 = [dh_dx, dh_dy]
print("2. Gradient vector [dh/dx, dh/dy]:")
print("∇h2 =", grad_h2)
print("-" * 40)

# 3. Evaluate at (x=2, y=0)
point = {x: 2, y: 0}
grad_eval = [item.subs(point) for item in grad_h2]

print("3. Gradient evaluated at (x=2, y=0):")
print("∇h2(2, 0) =", grad_eval)
```

```python
# -----------------------------------------------------------
# 🔹 3A. CHAIN RULE BY HAND vs SymPy
# -----------------------------------------------------------

# y = sin(x**2) is a composition: outer = sin(u), inner = u = x**2
# Chain rule:  dy/dx = cos(u) * du/dx = cos(x**2) * 2x
by_hand = sp.cos(x**2) * 2*x
by_sympy = sp.diff(sp.sin(x**2), x)

print('By hand :', by_hand)
print('By SymPy:', by_sympy)
print('Match?  ', sp.simplify(by_hand - by_sympy) == 0)
```

```python
# -----------------------------------------------------------
# 🔹 3B. CHAINING THREE FUNCTIONS
# -----------------------------------------------------------

# y = (3x + 1)**4  -> outer^4, inner (3x+1)
expr3 = (3*x + 1)**4
print('d/dx (3x+1)^4 =', sp.diff(expr3, x))   # 12*(3x+1)^3
```

```python
import sympy as sp

# Define symbolic variable
x = sp.Symbol("x")

# Definition of y = exp(x**2 + 1)
y = sp.exp(x**2 + 1)

# 1. By hand (outer derivative * inner derivative)
# inner = x**2 + 1, inner' = 2*x, outer' = exp(inner)
by_hand = sp.exp(x**2 + 1) * (2 * x)

print("1. Derivative by hand:")
print(by_hand)
print("-" * 40)

# 2. With sp.diff
by_sympy = sp.diff(y, x)

print("2. Derivative using sp.diff:")
print(by_sympy)
print("-" * 40)

# 3. Confirm they match
diff_check = sp.simplify(by_hand - by_sympy)
is_matching = diff_check == 0

print("3. Verification:")
print("Difference (by_hand - by_sympy):", diff_check)
print("Do the expressions match?:", is_matching)
```

```python
# -----------------------------------------------------------
# 🔹 4A. FORWARD PASS
# -----------------------------------------------------------

# Tiny toy problem: 4 samples, 3 input features, 5 hidden units, 1 output
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1

z1 = X @ W1                 # linear layer 1
h  = np.maximum(0, z1)      # ReLU activation
y_hat = h @ W2              # linear layer 2 (prediction)
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

```python
# -----------------------------------------------------------
# 🔹 4B. BACKWARD PASS (gradients via the chain rule)
# -----------------------------------------------------------

# Work backwards from the loss, one link at a time
dy   = 2 * (y_hat - Y) / Y.size      # d loss / d y_hat
dW2  = h.T @ dy                      # d loss / d W2
dh   = dy @ W2.T                     # d loss / d h
dz1  = dh * (z1 > 0)                 # ReLU gradient (1 where z1>0 else 0)
dW1  = X.T @ dz1                     # d loss / d W1

print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')
```

```python
# -----------------------------------------------------------
# 🔹 4C. ONE GRADIENT-DESCENT STEP SHOULD LOWER THE LOSS
# -----------------------------------------------------------

lr = 0.1
W1 -= lr * dW1               # step downhill
W2 -= lr * dW2

# Recompute the loss after the update
h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```

```python
import numpy as np

# Seed for reproducibility
np.random.seed(42)

# Provided inputs and weights
Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)
Wa = np.random.randn(4, 8) * 0.1  # input -> hidden (4x8)
Wb = np.random.randn(8, 1) * 0.1  # hidden -> output (8x1)

N = Xb.shape[0]  # Batch size = 6

# ---------------------------------------------------------------
# 1. Forward Pass
# ---------------------------------------------------------------
z1 = Xb @ Wa  # (6, 8)
h = np.maximum(0, z1)  # ReLU activation (6, 8)
y_hat = h @ Wb  # Linear output (6, 1)

# Mean Squared Error (MSE) Loss
loss_before = np.mean((y_hat - Yb) ** 2)

# ---------------------------------------------------------------
# 2. Backward Pass (Chain Rule)
# ---------------------------------------------------------------
# Loss = (1/N) * sum((y_hat - Yb)^2)
# dLoss/dy_hat = (2/N) * (y_hat - Yb)
dy = (2 / N) * (y_hat - Yb)  # (6, 1)

# dLoss/dWb = h.T @ dy
dWb = h.T @ dy  # (8, 1)

# dLoss/dh = dy @ Wb.T
dh = dy @ Wb.T  # (6, 8)

# Backprop through ReLU: dz1 = dh * (z1 > 0)
dz1 = dh * (z1 > 0)  # (6, 8)

# dLoss/dWa = Xb.T @ dz1
dWa = Xb.T @ dz1  # (4, 8)

# ---------------------------------------------------------------
# 3. Gradient Descent Step (lr = 0.05)
# ---------------------------------------------------------------
lr = 0.05
Wa -= lr * dWa
Wb -= lr * dWb

# Recalculate loss after update
z1_new = Xb @ Wa
h_new = np.maximum(0, z1_new)
y_hat_new = h_new @ Wb
loss_after = np.mean((y_hat_new - Yb) ** 2)

print(f"Loss before step: {loss_before:.6f}")
print(f"Loss after step:  {loss_after:.6f}")
```

```python
# -----------------------------------------------------------
# 🔹 5A. THE HESSIAN (matrix of second derivatives = curvature)
# -----------------------------------------------------------
x,y=sp.symbols('x y')
f5 = x**2 + 3*x*y + y**2
H = sp.hessian(f5, (x, y))
print('Hessian of f:')
sp.pprint(H)        # [[2, 3], [3, 2]]
```

```python
# -----------------------------------------------------------
# 🔹 5B. GRADIENT DESCENT ON A SIMPLE FUNCTION
# -----------------------------------------------------------

# Minimise f(x) = (x - 4)**2, whose minimum is at x = 4.
# Update rule:  x <- x - lr * f'(x),  with f'(x) = 2*(x - 4)
xv = 0.0          # starting point
lr = 0.2
for step in range(15):
    grad = 2 * (xv - 4)        # the derivative
    xv = xv - lr * grad        # step against the gradient
print('Converged x:', round(xv, 3), ' (true minimum = 4)')
```

```python
import sympy as sp

# ---------------------------------------------------------------
# 1. Hessian of x**4 + y**2
# ---------------------------------------------------------------
x, y = sp.symbols("x y")
f = x**4 + y**2

# Compute Hessian matrix (2x2 matrix of second-order partial derivatives)
H = sp.hessian(f, (x, y))

print("1. Hessian Matrix of x**4 + y**2:")
print(H)
print("-" * 45)

# ---------------------------------------------------------------
# 2. Gradient descent to minimize (x - 7)**2
# ---------------------------------------------------------------
xv = 0.0  # Initial starting point
lr = 0.1  # Learning rate

for step in range(20):
    # Derivative of f(x) = (x - 7)**2 is f'(x) = 2*(x - 7)
    grad = 2 * (xv - 7)

    # Gradient descent update rule: x = x - lr * grad
    xv = xv - lr * grad

print("2. Gradient Descent Results:")
print(f"Final x after 20 steps: {xv:.6f}")
print("Target optimal value:   7.0")
```
